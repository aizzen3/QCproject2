```python
# Cell 5 — Output folder and numerical settings


n_qubits = 8
Nx = 2 ** n_qubits

# Layer 0 means no circuit is applied.
# Layers 1–20 use the custom repeated bond-3 initializer.
layer_list = range(0, 15)

quimb_cutoff = 1e-16
plot_floor = 1e-18

T = 16e-6
lam = 630e-9
n_refr = 1.0

h = 1.0
sinc_scale = 4e6
x0 = 0.0


# Maximum bond dimensions you want to compare
max_bond_values = [3, 4, 20]

# Fidelity thresholds you want to request
fidelity_thresholds = [
    0.95,
    0.99,
    0.999,
]

# All cuts for 8 qubits:
# 1|7, 2|6, 3|5, 4|4, 5|3, 6|2, 7|1
cuts_to_plot = range(1, n_qubits)

# Set to 0.0 if you do not want singular-value cutoff
adaptive_cutoff = 0.0
quimb_cutoff = 1e-6


print("Number of qubits:", n_qubits)
print("Statevector length:", Nx)

```

```python
# Cell 6 — Signal functions

def spatial_grid(L, Nx):
    dx = L / Nx
    j = np.arange(Nx)
    return (j - (Nx - 1) / 2) * dx


def thickness_profile_sinc(x, h, scale=1.0, x0=0.0):
    z = scale * (x - x0)

    sinc = np.ones_like(z)
    nonzero = np.abs(z) > 1e-14
    sinc[nonzero] = np.sin(z[nonzero]) / z[nonzero]

    return h * sinc


def thickness_profile_positive_sinc(x, h, scale=1.0, x0=0.0):
    z = scale * (x - x0)

    sinc = np.ones_like(z)
    nonzero = np.abs(z) > 1e-14
    sinc[nonzero] = np.sin(z[nonzero]) / z[nonzero]

    sinc_positive = sinc - np.min(sinc)
    max_val = np.max(sinc_positive)

    if max_val > 0:
        sinc_positive = sinc_positive / max_val

    return h * sinc_positive


def phase_signal(d, n_refr, lam):
    k0 = 2 * np.pi / lam
    return n_refr * k0 * d


def phi_from_f(f):
    alpha = np.sum(f)

    if alpha <= 0:
        raise ValueError("alpha must be positive.")

    if np.any(f < 0):
        raise ValueError(
            "f contains negative values. Use the positive-sinc profile."
        )

    phi = np.sqrt(f / alpha)
    return phi, alpha

```

```python
# Cell 7 — Quimb and plotting helper functions



def dense_state_to_quimb_mps(psi, n_qubits, cutoff=1e-16, max_bond=None):
    psi = np.asarray(psi, dtype=complex)
    psi = psi / np.linalg.norm(psi)

    psi_tensor = psi.reshape([2] * n_qubits)

    return qtn.MatrixProductState.from_dense(
        psi_tensor,
        dims=[2] * n_qubits,
        cutoff=cutoff,
        max_bond=max_bond,
    )


def quimb_bond_report(qmps):
    bond_sizes = []

    for i in range(qmps.L - 1):
        bond = qmps.bond(i, i + 1)
        bond_sizes.append(qmps.ind_size(bond))

    max_bond = max(bond_sizes) if bond_sizes else 1
    return bond_sizes, max_bond


def quimb_singular_values(qmps, cut):
    """
    Return the singular values across the selected MPS cut.

    cut = 1 means:
        1 qubit | remaining qubits

    No manual cutoff is applied.
    """

    qmps_copy = qmps.copy()

    singular_values = qmps_copy.singular_values(cut)

    singular_values = np.asarray(
        singular_values,
        dtype=float,
    )

    # Keep descending order
    singular_values = np.sort(
        singular_values
    )[::-1]

    return singular_values


def pad_to_length(arr, length):
    arr = np.asarray(arr, dtype=float)

    output = np.zeros(length, dtype=float)
    retained_length = min(len(arr), length)
    output[:retained_length] = arr[:retained_length]

    return output

```

```python
# Cell 8 — Build the positive-sinc target state

x = spatial_grid(T, Nx)

d = thickness_profile_positive_sinc(
    x=x,
    h=h,
    scale=sinc_scale,
    x0=x0,
)

f = phase_signal(
    d=d,
    n_refr=n_refr,
    lam=lam,
)

psi, alpha = phi_from_f(f)

psi = np.asarray(psi, dtype=complex)
psi = psi / np.linalg.norm(psi)

target_state = Statevector(psi)

zero_state = np.zeros_like(psi, dtype=complex)
zero_state[0] = 1.0

print("alpha:", alpha)
print("n_qubits:", n_qubits)
print("Nx:", Nx)
print("norm psi:", np.linalg.norm(psi))
print("min |psi|:", np.min(np.abs(psi)))
print("max |psi|:", np.max(np.abs(psi)))

```

```python
all_results = []

psi_target = np.asarray(psi, dtype=complex)
psi_target = psi_target / np.linalg.norm(psi_target)


for max_bond in max_bond_values:

    for requested_fidelity in fidelity_thresholds:

        print("\n" + "=" * 75)
        print("Maximum search bond dimension:", max_bond)
        print("Requested fidelity:", requested_fidelity)
        print("=" * 75)

        try:
            result = adaptive_mps_from_statevector(
                psi_target,
                fidelity_threshold=requested_fidelity,
                max_search_dim=max_bond,
                cutoff=adaptive_cutoff,
                normalize=True,
            )

        except ValueError as error:
            print("Skipped:")
            print(error)
            continue

        U_psi = result["circuit"]
        report = result["report"]

        prepared_state, bond_leakage = (
            physical_statevector_from_mps_circuit(
                U_psi,
                n_physical=n_qubits,
                normalize=True,
            )
        )

        prepared_state = np.asarray(
            prepared_state,
            dtype=complex,
        )
        prepared_state /= np.linalg.norm(prepared_state)

        achieved_fidelity = float(
            np.abs(
                np.vdot(
                    psi_target,
                    prepared_state,
                )
            ) ** 2
        )

        qmps_prepared = dense_state_to_quimb_mps(
            psi=prepared_state,
            n_qubits=n_qubits,
            cutoff=0.0,
            max_bond=None,
        )

        bond_profile, actual_max_bond = (
            quimb_bond_report(qmps_prepared)
        )

        singular_values_by_cut = {}

        for cut in cuts_to_plot:
            singular_values_by_cut[cut] = (
                quimb_singular_values(
                    qmps=qmps_prepared,
                    cut=cut,
                )
            )

        all_results.append(
            {
                "max_search_dim": max_bond,
                "requested_fidelity": requested_fidelity,
                "achieved_fidelity": achieved_fidelity,
                "bond_leakage": bond_leakage,
                "bond_profile": bond_profile,
                "actual_max_bond": actual_max_bond,
                "singular_values_by_cut":
                    singular_values_by_cut,
                "circuit_depth": U_psi.depth(),
                "circuit_size": U_psi.size(),
                "report": report,
            }
        )

        print("Achieved fidelity:", achieved_fidelity)
        print("Quimb bond profile:", bond_profile)
        print("Quimb maximum bond:", actual_max_bond)
        print("Bond leakage:", bond_leakage)
```

```python
from qiskit.quantum_info import Statevector


def physical_chi_from_inverse_circuit(
    U_psi,
    psi,
    n_physical,
):
    """
    Construct the physical residual state

        chi = projection_bond=0 [U_psi† (|0_bond> tensor |psi>)]

    Assumption:
        Physical qubits are q0, ..., q(n_physical-1).
        Any bond/ancilla qubits are the higher-index qubits.
    """

    psi = np.asarray(psi, dtype=complex)
    psi = psi / np.linalg.norm(psi)

    total_qubits = U_psi.num_qubits
    n_extra = total_qubits - n_physical

    if n_extra < 0:
        raise ValueError(
            "The circuit has fewer qubits than the physical state."
        )

    # --------------------------------------------------
    # No additional bond qubits
    # --------------------------------------------------

    if n_extra == 0:

        full_target_state = Statevector(psi)

        full_chi_state = full_target_state.evolve(
            U_psi.inverse()
        )

        chi = np.asarray(
            full_chi_state.data,
            dtype=complex,
        )

        chi = chi / np.linalg.norm(chi)

        return chi, 1.0, 0

    # --------------------------------------------------
    # Add bond qubits in |00...0>
    #
    # ancilla_state.tensor(target_state) places psi on
    # the lower-index physical qubits and ancillas on
    # the higher-index qubits.
    # --------------------------------------------------

    ancilla_zero = np.zeros(
        2**n_extra,
        dtype=complex,
    )
    ancilla_zero[0] = 1.0

    full_target_state = Statevector(
        ancilla_zero
    ).tensor(
        Statevector(psi)
    )

    if full_target_state.num_qubits != total_qubits:
        raise ValueError(
            "Embedded target and circuit qubit counts do not match."
        )

    # Apply U†
    full_chi_state = full_target_state.evolve(
        U_psi.inverse()
    )

    full_chi = np.asarray(
        full_chi_state.data,
        dtype=complex,
    )

    # --------------------------------------------------
    # Select the bond-|00...0> block
    #
    # Since bond qubits are higher-index qubits, the
    # |00...0> sector is the first 2**n_physical entries.
    # --------------------------------------------------

    physical_dimension = 2**n_physical

    chi_unnormalized = full_chi[
        :physical_dimension
    ]

    projection_probability = float(
        np.linalg.norm(chi_unnormalized) ** 2
    )

    if projection_probability <= 0:
        raise ValueError(
            "Projection onto the zero-bond sector has zero norm."
        )

    chi = (
        chi_unnormalized
        / np.linalg.norm(chi_unnormalized)
    )

    return chi, projection_probability, n_extra
```

```python
max_bond_values = [3, 4, 20]

fidelity_thresholds = [
    0.95,
    0.99,
    0.999,
]

# No SVD cutoff in the adaptive approximation
adaptive_cutoff = 0.0

# No Quimb cutoff when converting dense states
quimb_cutoff = 1e-5

cuts_to_plot = range(1, n_qubits)

psi_target = np.asarray(
    psi,
    dtype=complex,
)

psi_target = (
    psi_target
    / np.linalg.norm(psi_target)
)

zero_state = np.zeros_like(
    psi_target,
    dtype=complex,
)

zero_state[0] = 1.0
```

```python
def quimb_singular_values(qmps, cut):
    """
    Return all numerical singular values across a cut.
    No manual cutoff is applied.
    """

    qmps_copy = qmps.copy()

    singular_values = np.asarray(
        qmps_copy.singular_values(cut),
        dtype=float,
    )

    return np.sort(singular_values)[::-1]
```

```python
all_results = []


for max_bond in max_bond_values:

    for requested_fidelity in fidelity_thresholds:

        print("\n" + "=" * 75)
        print("Maximum search bond:", max_bond)
        print("Requested fidelity:", requested_fidelity)
        print("=" * 75)

        # --------------------------------------------------
        # Adaptive MPS circuit
        # --------------------------------------------------

        try:

            result = adaptive_mps_from_statevector(
                psi_target,
                fidelity_threshold=requested_fidelity,
                max_search_dim=max_bond,
                cutoff=adaptive_cutoff,
                normalize=True,
            )

        except ValueError as error:

            print("Combination skipped:")
            print(error)
            continue

        U_psi = result["circuit"]
        report = result["report"]

        # ==================================================
        # A. Prepared physical state
        # ==================================================

        prepared_state, bond_leakage = (
            physical_statevector_from_mps_circuit(
                U_psi,
                n_physical=n_qubits,
                normalize=True,
            )
        )

        prepared_state = np.asarray(
            prepared_state,
            dtype=complex,
        )

        prepared_state = (
            prepared_state
            / np.linalg.norm(prepared_state)
        )

        achieved_fidelity = float(
            np.abs(
                np.vdot(
                    psi_target,
                    prepared_state,
                )
            ) ** 2
        )

        # Convert prepared state to Quimb MPS
        qmps_prepared = dense_state_to_quimb_mps(
            psi=prepared_state,
            n_qubits=n_qubits,
            cutoff=quimb_cutoff,
            max_bond=None,
        )

        prepared_bond_profile, prepared_max_bond = (
            quimb_bond_report(qmps_prepared)
        )

        prepared_singular_values_by_cut = {}

        for cut in cuts_to_plot:

            prepared_singular_values_by_cut[cut] = (
                quimb_singular_values(
                    qmps=qmps_prepared,
                    cut=cut,
                )
            )

        # ==================================================
        # B. Residual chi = U†|psi>
        # ==================================================

        chi, chi_projection_probability, n_extra = (
            physical_chi_from_inverse_circuit(
                U_psi=U_psi,
                psi=psi_target,
                n_physical=n_qubits,
            )
        )

        chi = np.asarray(
            chi,
            dtype=complex,
        )

        chi = chi / np.linalg.norm(chi)

        # Fidelity of chi with |000...0>
        chi_fidelity_zero = float(
            np.abs(
                np.vdot(
                    zero_state,
                    chi,
                )
            ) ** 2
        )

        # Convert chi to Quimb MPS
        qmps_chi = dense_state_to_quimb_mps(
            psi=chi,
            n_qubits=n_qubits,
            cutoff=quimb_cutoff,
            max_bond=None,
        )

        chi_bond_profile, chi_max_bond = (
            quimb_bond_report(qmps_chi)
        )

        chi_singular_values_by_cut = {}

        for cut in cuts_to_plot:

            chi_singular_values_by_cut[cut] = (
                quimb_singular_values(
                    qmps=qmps_chi,
                    cut=cut,
                )
            )

        # ==================================================
        # Store everything
        # ==================================================

        all_results.append(
            {
                "max_search_dim": max_bond,
                "requested_fidelity": requested_fidelity,

                "achieved_fidelity": achieved_fidelity,
                "prepared_state": prepared_state,
                "prepared_bond_profile":
                    prepared_bond_profile,
                "prepared_max_bond":
                    prepared_max_bond,
                "prepared_singular_values_by_cut":
                    prepared_singular_values_by_cut,

                "chi": chi,
                "chi_fidelity_zero":
                    chi_fidelity_zero,
                "chi_projection_probability":
                    chi_projection_probability,
                "chi_bond_profile":
                    chi_bond_profile,
                "chi_max_bond":
                    chi_max_bond,
                "chi_singular_values_by_cut":
                    chi_singular_values_by_cut,

                "bond_leakage": bond_leakage,
                "extra_qubits": n_extra,
                "circuit_depth": U_psi.depth(),
                "circuit_size": U_psi.size(),
                "report": report,
            }
        )

        print("Circuit total qubits:", U_psi.num_qubits)
        print("Extra bond qubits:", n_extra)

        print("\nPreparation result")
        print("Achieved fidelity:", achieved_fidelity)
        print(
            "Prepared bond profile:",
            prepared_bond_profile,
        )
        print(
            "Prepared max bond:",
            prepared_max_bond,
        )

        print("\nInverse result")
        print(
            "Fidelity chi with |000...0>:",
            chi_fidelity_zero,
        )
        print(
            "Bond-zero projection probability:",
            chi_projection_probability,
        )
        print("Chi bond profile:", chi_bond_profile)
        print("Chi max bond:", chi_max_bond)
```

```python
all_results = sorted(
    all_results,
    key=lambda row: row["achieved_fidelity"],
)

achieved_fidelities = np.asarray(
    [
        row["achieved_fidelity"]
        for row in all_results
    ],
    dtype=float,
)

print("\nSuccessful runs:")

for row in all_results:

    print(
        f"Dmax={row['max_search_dim']:2d}",
        f"| requested={row['requested_fidelity']}",
        f"| achieved={row['achieved_fidelity']:.12f}",
        f"| F(chi,0)={row['chi_fidelity_zero']:.12f}",
        f"| chi Dmax={row['chi_max_bond']}",
    )
```
